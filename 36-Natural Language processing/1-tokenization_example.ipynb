```python
import nltk
nltk.download('punkt_tab')
```

```python
corpus="""Hello Welcome, I am Akki.
Trying to learn natural language processing and become an expert at it.
"""
```

```python
print(corpus)
```

```python
## Tokenization
## Sentence-->paragraphs

from nltk.tokenize import sent_tokenize
```

```python
document=sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
  print(sentence)
```

```python
## Tokenization 
## Paragraphs-->words
## Paragraphs-->sentences

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
  print(word_tokenize(sentence))
```

```python
# full stop will be treated as separate
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
# full stop will not be treated as separate except for last sentence
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
