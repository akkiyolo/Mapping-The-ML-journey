##### Stemming
Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes to the roots of words known as lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).

##### RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

```python
## classification problem
## comments of products is a positive review or negative review
## reviews---> eating , eaten , written, writing , writes
```

```python
## porterstemmer

from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
words=["eating","eaten","programming","programs","finally","finalized","writing","writes"]
```

```python
for word in words:
  print(word+"--->"+stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem("sitting")
```

```python
## regexpstemmer class

from nltk.stem import RegexpStemmer
```

```python
# what it says is that if there is an ing,s,e,able then remove that
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$',min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

#### Snowball Stemmer
The Snowball Stemmer is an improved and more efficient version of the classic Porter Stemmer. It is designed for better accuracy, speed, and multilingual support in NLP tasks.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer=SnowballStemmer('english')
```

```python
for word in words:
  print(word+"--->"+snowballstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballstemmer.stem("fairly"),snowballstemmer.stem("sportingly")
```

```python
snowballstemmer.stem('goes')
```
