#### Missing Values

##### Missing values occur in a dataset when some of the information is not stored in a variable , there are  mechanisms:

### 1. MCAR — Missing Completely At Random

#### Meaning:

The missingness has nothing to do with the data itself — neither the observed values nor the unobserved ones.

👉 In other words, the reason for missing data is purely random.

#### Example:

Suppose you are collecting data about students' exam scores, and some answer sheets are lost accidentally in a fire.
The missing values have nothing to do with the student’s performance, gender, or age — they’re just gone randomly.

### 2. MAR — Missing At Random

#### Meaning:

The missingness is related to some other observed variable, but not to the variable itself that has missing data.

👉 The missing pattern can be explained using other variables.

#### Example:

In a medical survey:

Older people are less likely to report their income.
So, the missingness in Income depends on Age, which is known.

### 3. MNAR — Missing Not At Random
#### Meaning:

The missingness depends on the unobserved data itself — i.e., the value that’s missing affects the fact that it’s missing.

👉 The reason for missing is not random and not explainable by other data.

#### Example:

People with very high income may choose not to disclose their income.

Patients with severe illness may drop out of a medical study.

Here, the missingness depends on the value itself (income, illness severity).
